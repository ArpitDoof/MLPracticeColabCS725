```python
from sklearn.tree import DecisionTreeClassifier
```

```python
x = iris.data[:,2:]
```

```python
y = iris.target
```

```python
tree_clf = DecisionTreeClassifier(max_depth =2) #maxDepth is. hyper parameter
tree_clf.fit(x,y)
```

```python
#This code trains the decision tree classifier over the Iris data set
```

```python
from sklearn.tree import export_graphviz
```

```python
export_graphviz(
    tree_clf,
    out_file = "iris_tree.dot",
    feature_names = iris.feature_names[2:],
    class_names= iris.target_names,
    rounded= True,
    filled = True
)
```

```python
from graphviz import Source
```

```python
Source.from_file("iris_tree.dot")
```

```python
arr = np.array(tree_clf.predict_proba([[5,1.5]]))
```
